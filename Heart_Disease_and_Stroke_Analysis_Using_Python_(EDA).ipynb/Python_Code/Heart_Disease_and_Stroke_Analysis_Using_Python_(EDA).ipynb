Stage 5 — Data Cleaning

Step 1 – Handle Missing Values

```python
df.isnull().sum()
```

```python
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
df['Education'].fillna(df['Education'].mode()[0], inplace=True)
```

```python
df['Cigarettes_Per_Day'].fillna(df['Cigarettes_Per_Day'].median(), inplace=True)
```

```python

df['Total_Cholesterol'].fillna(df['Total_Cholesterol'].median(), inplace=True)
```

```python
df['Body_Mass_Index'].fillna(df['Body_Mass_Index'].median(), inplace=True)
```

```python
df['Heart_Rate'].fillna(df['Heart_Rate'].median(), inplace=True)
```

```python
df['Glucose'].fillna(df['Glucose'].median(), inplace=True)
```

```python
df.isnull().sum()
```

Step 2 – Rename the Blank Column

```python
df.rename(columns={
    df.columns[5]: 'BP_Medication'
}, inplace=True)
```

```python
print(df.columns.tolist())
```

```python
df['BP_Medication'] = df['BP_Medication'].fillna(
    df['BP_Medication'].mode()[0]
)
```

```python
df['BP_Medication'].isnull().sum()
```

```python
# Export cleaned dataset to Excel
df.to_excel('cleaned_heart_disease_dataset.xlsx',index=False)
print("Cleaned Dataset Exported Successfully!")
```

Stage 14 – Comparative Analysis.

1. Average Age by Heart Disease Status

```python
age_comparison = df.groupby('Heart_Disease_Status')['Age'].mean()

print(age_comparison)
```

2. Average Cholesterol by Heart Disease Status

```python
cholesterol_comparison = df.groupby('Heart_Disease_Status')['Total_Cholesterol'].mean()

print(cholesterol_comparison)
```

3. Average Blood Pressure by Heart Disease Status

```python
bp_comparison = df.groupby('Heart_Disease_Status')['Systolic_BP'].mean()

print(bp_comparison)
```

4. Average BMI by Heart Disease Status

```python
bmi_comparison = df.groupby('Heart_Disease_Status')['Body_Mass_Index'].mean()

print(bmi_comparison)
```

5. Compare Multiple Variables

```python
comparison = df.groupby('Heart_Disease_Status')[
    ['Age', 'Total_Cholesterol', 'Systolic_BP',
     'Diastolic_BP', 'Body_Mass_Index', 'Glucose']
].mean()

print(comparison)
```

6. Visualise the Comparison

```python
comparison.T.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Average Health Measurements by Heart Disease Status')
plt.xlabel('Health Measurements')
plt.ylabel('Average Value')
plt.xticks(rotation=45)
plt.legend(title='Heart Disease Status')
plt.tight_layout()
plt.show()
```
